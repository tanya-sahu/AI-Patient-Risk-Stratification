```python
import pandas as pd
import numpy as np
df = pd.read_csv("diabetes.csv")
print(df.shape)
df.head()
```

```python
cols = ["Glucose","BloodPressure","SkinThickness","Insulin","BMI"]

for col in cols:
    df[col] = df[col].replace(0, df[col].median())
df.head()    
```

```python
df.to_csv("final_diabetes.csv", index=False)
```

```python
X = df.drop("Outcome" , axis=1)
Y = df["Outcome"]
Y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , Y_train , Y_test = train_test_split(X, Y , test_size = 0.2 , random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(min_samples_split = 5 , min_samples_leaf  = 6 , 
                            n_estimators = 200 , max_samples = 0.50 , random_state= 42 , max_depth = 7, bootstrap = True, oob_score = True ,    class_weight='balanced',)
rf.fit(X_train , Y_train)
Y_pred = rf.predict(X_test)
Y_pred
```

```python
from sklearn.metrics import confusion_matrix , accuracy_score , classification_report
cf = confusion_matrix(Y_test , Y_pred)
cf
```

```python
accuracy_score(Y_test , Y_pred)*100
```

```python
print(classification_report(Y_test, Y_pred))
```

```python
import matplotlib.pyplot as plt

importances = rf.feature_importances_
features = X.columns

importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': importances
})

importance_df = importance_df.sort_values(by='Importance', ascending=False)

plt.barh(importance_df['Feature'], importance_df['Importance'])
plt.xlabel("Importance")
plt.ylabel("Features")
plt.title("Feature Importance (Random Forest)")
plt.show()
```

```python

# meshgrid
x_min, x_max = df["BMI"].min()-1, df["BMI"].max()+1
y_min, y_max = df["Glucose"].min()-1, df["Glucose"].max()+1

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 1),
    np.arange(y_min, y_max, 1)
)

# create full feature input (IMPORTANT)
grid = np.c_[
    np.zeros(len(xx.ravel())),          # Pregnancies
    yy.ravel(),                         # Glucose
    np.full(len(xx.ravel()), 80),       # BloodPressure
    np.full(len(xx.ravel()), 20),       # SkinThickness
    np.full(len(xx.ravel()), 80),       # Insulin
    np.full(len(xx.ravel()), 25),       # BMI
    np.full(len(xx.ravel()), 0.5),      # DiabetesPedigreeFunction
    xx.ravel()                          # Age
]

# prediction (model already trained)
Z = rf.predict(grid)
Z = Z.reshape(xx.shape)

# plot
plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(df["BMI"], df["Glucose"], c=df["Outcome"], edgecolor='k')

plt.xlabel("BMI")
plt.ylabel("Glucose")
plt.title("Decision Boundary (BMI vs Glucose)")
plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

x = df["Age"]
y = df["BMI"]
z = df["Glucose"]
c = df["Outcome"]

scatter = ax.scatter(x, y, z, c=c)

ax.set_xlabel("Age")
ax.set_ylabel("BMI")
ax.set_zlabel("Glucose")
ax.set_title("3D Visualization (Age vs BMI vs Glucose)")

plt.show()
```

```python
import plotly.express as px
import numpy as np

# base data
C = df[["Age", "BMI", "Glucose"]]

# baaki features ko fix kar do
def create_input(age, bmi, glucose):
    return [
        0,              # Pregnancies
        glucose,
        80,             # BP
        20,             # SkinThickness
        80,             # Insulin
        bmi,
        0.5,            # DPF
        age
    ]

# model prediction on actual data
probs = []

for i in range(len(df)):
    row = create_input(df["Age"][i], df["BMI"][i], df["Glucose"][i])
    prob = rf.predict_proba([row])[0][1]
    probs.append(prob)

df["Risk"] = probs

# 3D plot
fig = px.scatter_3d(
    df,
    x="Age",
    y="BMI",
    z="Glucose",
    color="Risk",   # model output
    title="3D Model Decision Visualization"
)

fig.show()
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100,200],
    'max_depth': [3,5,7],
    'min_samples_split': [2,5]
}

grid = GridSearchCV(RandomForestClassifier(), param_grid, cv=5)
grid.fit(X_train, Y_train)

print(grid.best_params_)
```

```python
from sklearn.feature_selection import SelectFromModel


rf.fit(X_train, Y_train)

selector = SelectFromModel(rf, threshold=0.02)

X_train_selected = selector.fit_transform(X_train, Y_train)
X_test_selected = selector.transform(X_test)


selected_features = X_train.columns[selector.get_support()]

print(selected_features)
```

```python
After handling class imbalance, false negatives reduced significantly,
and recall improved from around 65% to 82%.

Although accuracy slightly decreased, the model is now more reliable
for healthcare use as it misses fewer patients.
```
