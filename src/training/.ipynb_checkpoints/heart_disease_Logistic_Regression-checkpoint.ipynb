```python
## Dataset Reading
import pandas as pd
import numpy as np
df = pd.read_csv("processed_heart_disease_uci.csv")
df
```

```python
from sklearn.model_selection import train_test_split
columns = df.columns
columns = columns.drop("target")
```

```python
X = df[columns]
X.shape
```

```python
Y = df["target"]
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , Y , test_size = 0.2 , random_state= 42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
## Scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train , y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
y_pred = lr.predict(X_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix , precision_score , recall_score , f1_score , accuracy_score
confusion_matrix(y_pred , y_test)
confusion_matrix(y_test , y_pred)
```

```python
precision_score(y_pred , y_test)*100
```

```python
recall_score(y_pred , y_test)*100
```

```python
accuracy_score(y_pred , y_test)*100
```

```python
precision_score(y_pred , y_test , average=None)
```

```python
precision_score(y_pred , y_test , average=None).shape
```

```python
recall_score(y_pred , y_test, average=None)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
